```python
import json, re
```

```python
textbook_file = '../data/raw_text/Exercises_in_latin_prosody/Exercises_in_Latin_Prosody_and_Versifica.txt'
key_file = '../data/raw_text/Exercises_in_latin_prosody/A_Key_to_Exercises_in_Latin_Prosody_and.txt'

with open(textbook_file, 'r') as f:
    textbook = f.read()

with open(key_file, 'r') as f:
    answer_key = f.read()
```

```python
# get rid of intro text in textbook
_, textbook = textbook.split('viii \nPREFACE.')
print(textbook[:50])
```

```python
# break into parts
textbook_parts = re.split('(PART [IVX]+)', textbook)
textbook_parts = textbook_parts[1:] # get rid of text before first part

key_parts = re.split('(PART [IVX]+)', answer_key)
key_parts = key_parts[1:]
```

```python
key_parts
```

```python
# convert to dictionaries
textbook_dict = {}
for i in range(0, len(textbook_parts), 2):
    part_name = textbook_parts[i]
    part_text = textbook_parts[i+1]

    textbook_dict[part_name] = part_text

key_dict = {}
for i in range(0, len(key_parts), 2):
    part_name = key_parts[i]
    part_text = key_parts[i+1]

    key_dict[part_name] = part_text
```

```python
# split into chapters
def chapter_split(part_text) -> dict:
    chapters = re.split('(CHAPTER [IVX]+)', part_text)
    chapters = chapters[1:] # remove text before chapter

    # turn into dict
    assert len(chapters) % 2 == 0
    chapters_dict = {}
    for i in range(0, len(chapters), 2):
        chapter_name = chapters[i]
        chapter_text = chapters[i+1]

        chapters_dict[chapter_name] = chapter_text
    
    return chapters_dict
```

```python
chapter_split(textbook_dict["PART I"])
```

```python
for part_name, part_text in textbook_dict.items():
    chapters_dict = chapter_split(part_text)
    textbook_dict[part_name] = chapters_dict

for part_name, part_text in key_dict.items():
    chapters_dict = chapter_split(part_text)
    key_dict[part_name] = chapters_dict

textbook_dict

```

```python
# get exercise text from textbook
for part_name, chapter_dict in textbook_dict.items():
    for chapter_name, chapter_text in chapter_dict.items():
        print(chapter_name)
        print(chapter_text)
        print('--------')
        _, *exercise_texts = chapter_text.split("EXERCISES.")
        textbook_dict[part_name][chapter_name] = '\n'.join(exercise_texts)
```

```python
# save to json
with open('../data/semi_structured/prosody_questions.json', 'w') as f:
    json.dump(textbook_dict, f, indent=4)

with open('../data/semi_structured/prosody_key.json', 'w') as f:
    json.dump(key_dict, f, indent=4)
```
