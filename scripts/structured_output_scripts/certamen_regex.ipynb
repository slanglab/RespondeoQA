```python
import re, json
```

```python
infile = '../data/raw_text/NJCL Certamen 1996-2009 Word Format.docx.txt'
with open(infile, 'r') as f:
    raw_txt = f.read()
```

```python
# separate by [YEAR] NCLJ [LEVEL] DIVISION
sections = re.split(r'(\d{4})\.? NJCL CERTAMEN.*?', raw_txt)
```

```python
len(sections)
```

```python
sections
```

```python
sections.insert(0, '1996')
```

```python
data = {}
for i in range(0, len(sections), 2):
    year = sections[i]
    section_text = sections[i+1]
    
    # search for NCLJ CERTAMEN NOVICE/LOWER/UPPER
    division = re.search(r'(NOVICE|LOWER|UPPER)', section_text)
    if division:
        division = division.group(1)
    else:
        division = 'UNKNOWN'

    key = f'{year}_{division}'
    questions = re.split(r'(?m)^\s*(?=\d+\.\s+)', section_text)
    
    data[key] = questions
```

```python
data
```

```python
with open('../data/semi_structured/certamen_1996_2009.json', 'w') as f:
    json.dump(data, f, indent=4)
```
