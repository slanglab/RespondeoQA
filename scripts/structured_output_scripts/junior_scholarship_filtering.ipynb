```python
import json
import pandas as pd

from openai import OpenAI
from time import sleep


with open('openai-api-key.txt', 'r') as f:
    openai_api_key = f.read().strip()

openai_client = OpenAI(api_key=openai_api_key)
```

```python
# separate the translation from other questions
with open('../data/structured/junior_scholarship.json', 'r') as f:
    questions = json.load(f)

question_df = pd.DataFrame(questions)
len(question_df)
```

```python
question_df['answers'].apply(type).value_counts()
```

```python
# drop empty answers (where answers cell is != [""])
question_df = question_df[question_df['answers'].apply(lambda x: x != [""])]
len(question_df)
```

```python
# unique question ids
len(question_df['question_id'].unique())

```

```python
# non-translation questions
short_answer_long = question_df[question_df['question_content'] != 'translation']
short_answer_long
```

```python
# filter to one word responses
short_answer = short_answer_long[short_answer_long['answers'].apply(len) == 1]
short_answer = short_answer[short_answer['answers'].apply(lambda x: ' ' not in x[0])]
short_answer
```

```python
short_answer_df = short_answer 

short_answer = short_answer_df.to_dict(orient='records')
short_answer
```

are short answer questions answerable?

```python
answerable_prompt='''I will give you a question and its answer.
Please decide if the question is answerable based on the given answer, and the added context that the question is related to Roman and Greek history, mythology, and geography, and Latin language and grammar. This additional context is provided along with the question, so you should take it into account.
Unanswerable questions don't have enough information. For example, if the question references a diagram, map, or table that is not provided, the question is unanswerable. If the question refers to a phrase or text that is not provided, the question is unanswerable. 

Please respond simply with "answerable" or "unanswerable".
'''
```

```python
len(short_answer)
```

```python
ids = []
responses = []

N = len(short_answer)

# continue from where we left off, or from first error response
for i in range(len(short_answer)):
    #if i < len(responses) and responses[i] != 'error':
    #    continue
    print(i, end=' ')

    q_id = short_answer[i]['question_id']
    question = short_answer[i]['question_text']
    answer = short_answer[i]['answers'][0]
    multiple_choice = short_answer[i]['multiple_choice_options']

    prompt = answerable_prompt 
    prompt += f"\nHere is the question: {question}\n"
    if multiple_choice:
        choice_str = '\n'.join(multiple_choice)
        prompt += f"Here are the multiple choice options: {choice_str}\n"
    prompt += f"Answer: {answer}\n"
    
    try:
        response = openai_client.chat.completions.create(
                model="gpt-4o",
                messages=[
                    {"role": "system", "content": "You are a helpful assistant."},
                    {"role": "user", "content": prompt}
                ]
            )
        resp = response.choices[0].message.content
    except Exception as e:
        print(f"Error on question {i}: {e}")
        resp = 'error'
    
    #if responses[i] == 'error':
    #    responses[i] = resp
    #elif i < len(responses):
    responses.append(resp)
    ids.append(q_id)

    if i % 50 == 0:
        print(f"{i+1}/{N}")
        # dump to json
        with open('../data/structured/junior_answerable_responses.json', 'w') as f:
            json.dump(responses, f, indent=4)

    sleep(.01)

# save final 
with open('../data/structured/junior_answerable_responses.json', 'w') as f:
    json.dump(responses, f, indent=4)
```

```python
from collections import Counter
lower_resp = [r.lower() for r in responses]
lower_resp = [r.replace('.', '') for r in lower_resp]
count_resp = Counter(lower_resp)
count_resp
```

```python
# load edited unanswerable questions
unanswerable_df = pd.read_csv('../data/structured/junior_unanswerable_edits.tsv', sep='\t')
unanswerable_df = unanswerable_df.drop(columns=['Unnamed: 0'])
unanswerable_df
```

```python
unanswerable_ids = unanswerable_df[unanswerable_df['answerable'] == 0]['question_id'].tolist()
len(unanswerable_ids)
```

```python
# ids where edit = 1
update_ids = unanswerable_df[unanswerable_df['edits'] == 1.0]['question_id'].tolist()
len(update_ids)
```

```python
# drop unanswerable from short_answer_df
short_answer_df = short_answer_df[~short_answer_df['question_id'].isin(unanswerable_ids)]
len(short_answer_df)
```

```python
# update the questions with edits
short_answer_dict = short_answer_df.to_dict(orient='records')
for i, dict_ in enumerate(short_answer_dict):
    this_id = dict_['question_id']
    if this_id in update_ids:
        # find the row in unanswerable_df
        new_row = unanswerable_df[unanswerable_df['question_id'] == this_id]
        # drop the "edits" column
        new_row = new_row.drop(columns=['edits'])
        # turn into dict
        new_row = new_row.to_dict(orient='records')[0]
        # update the row
        short_answer_dict[i] = new_row
    # delete the "answerable" key 
    del short_answer_dict[i]['answerable']

short_answer_dict
```

```python
len(short_answer_dict)
```

```python
# save to json
with open('../data/structured/junior_scholarship_short_answer.json', 'w') as f:
    json.dump(short_answer_dict, f, indent=4)
```

```python
# load edited long translation questions 
with open('../data/structured/junior_scholarship_translation_edits.tsv', 'r') as f:
    edited_df = pd.read_csv(f, sep='\t')

edited_df.drop(columns=['Unnamed: 0'], inplace=True)
edited_df

```

```python
# drop rows where answerable = 0 (keep nan or 1)
edited_df = edited_df[edited_df['answerable'] != 0]
edited_df
```

```python
# drop the "answerable" and "edits" columns
edited_df = edited_df.drop(columns=['answerable', 'edits'])
edited_dict = edited_df.to_dict(orient='records')
edited_dict

```

```python
# make sure "answers" and "multiple_choice_options" are lists
for i, dict_ in enumerate(edited_dict):
    if isinstance(dict_['answers'], str):
        edited_dict[i]['answers'] = eval(dict_['answers'])
    if isinstance(dict_['multiple_choice_options'], str):
        edited_dict[i]['multiple_choice_options'] = eval(dict_['multiple_choice_options'])
edited_dict
```

```python
# do some cleanup of answer strings 
for i, dict_ in enumerate(edited_dict):
    answers = dict_['answers']
    for j, answer in enumerate(answers):
        answers[j] = answer.strip()
        answers[j] = answers[j].replace('*', '')

        # if " in the answer twice (start and end), remove both;
        # but otherwise leave them - might be a quote 
        if answer.count('"') == 2 and answer[0] == '"' and answer[-1] == '"':
            answers[j] = answer[1:-1]

    dict_['answers'] = answers

edited_dict
```

```python
# save to json
with open('../data/structured/junior_scholarship_translation_long.json', 'w') as f:
    json.dump(edited_dict, f, indent=4)
```
