```python
# for each page, want to extract the question blocks
# these will be the blocks of text that start with "1.\n", "2.\n", "3.\n", etc.def extract_question_blocks(text):

key_blocks = []
for page_num, key_text in enumerate(key_texts):
    # find the question blocks
    # these will be the blocks of text that start with "1.\n", "2.\n", "3.\n", etc.
    question_blocks = re.split(r'(\d+\.\n)', key_text)
    key_blocks.extend(question_blocks)

text_blocks = []
for page_num, text_text in enumerate(text_texts):
    # find the question blocks
    # these will be the blocks of text that start with "1.\n", "2.\n", "3.\n", etc.
    question_blocks = re.split(r'(\d+\.\n)', text_text)
    text_blocks.extend(question_blocks)

```

```python
# merge the blocks
# looking for numbers 1-24 only

def merge_text_blocks(text_blocks):
    merged_blocks = []
    curr_num = 0 # number we are currently adding to 
    curr_block = ""
    for block in text_blocks:
        if block.startswith(f"{curr_num + 1}.\n"): # we are done with curr num
            merged_blocks.append(curr_block)
            curr_block = block
            curr_num +=1
        else:
            curr_block += block

    # add the last block
    merged_blocks.append(curr_block)
    return merged_blocks



```

```python
exercise_instructions = 'Form the following line into hexameter or pentameter verse by changing the position of one word.'

all_italicized_instructions = 'The word whose position needs to be changed is italicized.'
```

```python
from token import NEWLINE


def clean_lines(lines):
    exclude_lines = ['', 'Lat. Pr.', 'E', '50', 'EXERCISES IN','12', 'CESURA.','13']
    new_lines = []
    for line in lines:
        if 'VERSIFICATION' in line:
            continue
        elif 'ARRANGEMENT' in line:
            continue 
        elif 'Digitized' in line: continue
        elif line in exclude_lines: continue
        new_lines.append(line)

    return new_lines
```

```python
final_exercises = []

for question, answer in zip(exercises, exercises_answers):
    # get the number 
    num = question.split('.')[0] 
    question_format = 'short_answer'
    question_content = 'scansion'
    difficulty = 'unknown'

    # strip the number and the period
    rm = f'{num}.\n'
    question = question[len(rm):]
    answer = answer[len(rm):]

    # also rm first line of answer, because it's a citation
    first_newline_idx = answer.find('\n')
    answer = answer[first_newline_idx+1:]

    q_lines = question.split('\n')
    a_lines = answer.split('\n')

    q_lines = clean_lines(q_lines)
    a_lines = clean_lines(a_lines)

    if len(q_lines) != len(a_lines):
        print(q_lines)
        print(a_lines)
        break

    for j, (q_line, a_line) in enumerate(zip(q_lines, a_lines)):
        q_id = f"prosody_{part_name}.{chap_name}.{num}.{j+1}"

        # add the instructions to the start
        # check how many italicized words there are by counting number of *
        num_italicized = q_line.count('*')
        if num_italicized == 0:
            question = exercise_instructions + '\n' + q_line
        elif num_italicized == 2:
            question = exercise_instructions + '\n' + all_italicized_instructions + '\n' + q_line
        else:
            print(f"Unexpected number of italicized words: {num_italicized}")
            print(question)
            print(answer)
            break
        
        final_exercises.append({
            'source_name': "exercises-in-latin-prosody",
            'source_year': 1823,
            'question_id': q_id,
            'question_format': 'short_answer',
            'question_content': 'scansion',
            'difficulty': 'unknown',
        
            'question_language': 'english',
            'answer_language': 'latin',

            'question_text': question,
            'multiple_choice_options': [],
            'answers': [a_line],

        })
```

```python
with open('../data/structured/prosody_caesura.json', 'w') as f:
    json.dump(final_exercises, f, indent=4)
```

```python
PROMPT_TEMPLATE = """I will give you a short list of questions and their answer key. You should separate them into individual questions and answers.
Please clean up any formatting issues, and do not include headers or other text that's not part of the question or answer.
Please format your answer like so:

1. question? (all one line)
answer (all one line)
2. question?
answer

Here are the questions:
{}

Here are the answers:
{}
"""
```

```python
from openai import OpenAI
from time import sleep


with open('openai-api-key.txt', 'r') as f:
    openai_api_key = f.read().strip()

openai_client = OpenAI(api_key=openai_api_key)
```

```python
assert len(short_answer_questions) == len(short_answers)

combined_strs = []
raw_responses = []
for question, answer in zip(short_answer_questions, short_answers):
    print(question)

    prompt = PROMPT_TEMPLATE.format(question, answer)

    response = openai_client.responses.create(
        model="gpt-4o",
        input=prompt
    )

    raw_responses.append(response)
    sleep(1)
    print('------------')

    
```
