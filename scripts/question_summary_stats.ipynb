final dataset summary stats

```python
import glob
import os
import json
```

```python
data_dir = 'data/final_dataset'
files = glob.glob(os.path.join(data_dir, '*.json'))
# remove reading comp passage files
files = [f for f in files if 'reading_comp_passages' not in f]

all_data = []
for file in files:
    with open(file, 'r') as f:
        all_data += json.load(f)
```

```python
len(all_data)
```

```python
all_df = pd.DataFrame(all_data)
all_df
```

```python
# summary of columns (source, question_content, question_language, answer_language)

cols = ['source_name', 'question_content', 'question_language', 'answer_language']
for col in cols:
    print(all_df[col].value_counts())

```

```python
# of x questions, whats their source?
types = all_df['question_content'].unique()
for type_ in types:
    print(type_)
    print(all_df[all_df['question_content'] == type_].source_name.value_counts())
    print()
```
